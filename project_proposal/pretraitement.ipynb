## Traitement des données manquantes en Python
Elle suit deux (02) étapes :
- Identification du type de valeurs manquantes
- Traitement de ces données.
Identification dy type de valeurs manquantes 
- MCAR (Complètement aléatoire)
Les données manquantes n'ont aucune realtion avec les autres données
Exemple de situation : Perte accidentelle de données
Méthodes de traitement recommandées :
    - Suppression des observations (si proportion faible),  
    - Imputation par la moyenne, la médiane
- MAR (Missing At Random) : Les valeurs manquantes dépendent des **données observées**, mais pas des données manquantes elle-même 
Exemple : Non-réponse liée à une variable observée
Traitement : Méthodes d'imputation (KNN, MissForest régression)
- MNAR (Non aléatoire)
    - Les données manquantes **dépendantes** des valeurs manquantes elles-mêmes. 
    Exemple : Il y a certaines personnes qui sont réticentes à partager leur revenue lors d'une enquête. Ainsi la valeur manquante peut-être dû à ce phénomène.
